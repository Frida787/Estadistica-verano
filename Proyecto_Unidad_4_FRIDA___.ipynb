## METODOLOGIA

Dimensión Fractal (Fractal Dimension)

La dimensión fractal es una medida que describe la **complejidad geométrica** de una estructura. A diferencia de las dimensiones euclidianas convencionales, la dimensión fractal puede tener valores no enteros, lo que permite caracterizar formas irregulares y auto-similares como los contornos tumorales. En el contexto oncológico, una **mayor dimensión fractal suele estar asociada con bordes más irregulares**, lo que se ha relacionado con una mayor probabilidad de malignidad. Esta medida es útil para cuantificar qué tan rugoso o caótico es el perímetro del tumor.

Compacidad (Compactness)

La compacidad es una métrica que evalúa **cuán cercana es la forma del tumor a un círculo perfecto**. Se calcula generalmente como una función del perímetro y el área de la figura, y valores más bajos indican una forma más irregular o dispersa, mientras que valores altos reflejan una figura más redondeada y homogénea. En términos clínicos, tumores con baja compacidad suelen ser más sospechosos, ya que la forma irregular puede reflejar un crecimiento anómalo.

Relación entre Dimensión Fractal y Compacidad

Ambas características están relacionadas con la **morfología del tumor**, y se ha planteado que podría existir una correlación entre ellas. Tumores con bordes muy complejos (alta dimensión fractal) tienden a mostrar formas menos regulares (baja compacidad). Analizar esta relación mediante métodos estadísticos puede ofrecer **indicadores adicionales para la clasificación de tumores** y aportar valor en sistemas de diagnóstico asistido por computadora (CAD).

```python
import pandas as pd

df= pd.read_csv("https://raw.githubusercontent.com/Frida787/Estadistica-verano/refs/heads/main/Cancer_Data.csv")
df=df.drop(columns=["Unnamed: 32", "id"])
df
```

```python
from statsmodels.formula.api import ols
import statsmodels.api as sm

#Y ~ X
modelo_lineal = ols("fractal_dimension_worst~C(diagnosis)",data=df).fit()
tabla_anova= sm.stats.anova_lm(modelo_lineal)
tabla_anova
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

nivel_de_significancia = 0.05

# Prueba de Tukey
tukey = pairwise_tukeyhsd(endog=df['fractal_dimension_worst'], groups=df['diagnosis'], alpha=nivel_de_significancia)

# Mostrar los resultados
print(tukey)

# Gráfico de las diferencias entre grupos
tukey.plot_simultaneous(ylabel="Grupos", xlabel="Medias")

plt.gca().spines['right'].set_visible(False) # derecha
plt.gca().spines['top'].set_visible(False)   # superior
plt.title("Múltiples comparaciones entre todos los pares (Tukey)")
```

```python
x = df["fractal_dimension_worst"]
y = df["compactness_worst"]


# Recta de regresión ajustada y grafíquelo sobre el gráfico de
# dispersión.
import statsmodels.api as sm
x_constante = sm.add_constant(x)
modelo = sm.OLS(y, x_constante).fit()

yc = modelo.predict(x_constante)
plt.figure()
plt.plot(y, y, color = 'gray', linestyle = '--')

# Coeficiente de determinación.
from sklearn.metrics import r2_score  # recomendada
r2 = r2_score(y, yc)
print(f'Coeficiente de determinación: {r2: 0.4f}\n')

# Intervalo de confianza del 95% para la pendiente de la recta de
# regresión ajustada ( b1 )
nivel_de_confianza = 0.95
intervalo_de_confianza = modelo.conf_int(alpha = 1 - nivel_de_confianza)
intervalo_de_confianza_b1 = intervalo_de_confianza.iloc[1]
print(f'Intervalo de confianza para b1 de {nivel_de_confianza: 0.0%}')
print(f'{intervalo_de_confianza_b1[0]: 0.4f} < b1 < {intervalo_de_confianza_b1[1]: 0.4f}\n')

# Residuales y trace un nuevo gráfico de dispersión. Comente,
# ¿Parece que se verifican los supuestos?
residuales = modelo.resid
plt.scatter(y, yc, color = 'blue')
plt.xlabel('Variable dependiente real')
plt.ylabel('Variable dependiente calculada')
plt.gca().spines[['top', 'right']].set_visible(False)

# Prueba de Shapiro para los residuales y comente el resultado.
from scipy.stats import shapiro
_, valor_p_sh = shapiro(residuales)
print(f'valor-p de Shapiro: {valor_p_sh: 0.4f}\n')

# Prueba de Brausch-Pagan para los residuales y comente el
# resultado.
from statsmodels.stats.api import het_breuschpagan
_, valor_p_bp, _, _ = het_breuschpagan(residuales, x_constante)
print(f'valor_p de Breusch-Pagan: {valor_p_bp: 0.4f}\n')


# Tabla ANOVA.
from statsmodels.formula.api import ols
# Y ~ X
modelo_lineal = ols('compactness_worst ~ fractal_dimension_worst', data = df).fit()
tabla_anova = sm.stats.anova_lm(modelo_lineal)
tabla_anova
```

##Conclusion

Se analizó la relación entre la variable dimensión fractal (como variable independiente) y la compacidad peor (como variable dependiente), con el objetivo de determinar la fuerza y características de su asociación mediante un modelo de regresión lineal simple.
1. Relación lineal y capacidad explicativa
El coeficiente de determinación (R²) fue de 0.6568, lo que indica que aproximadamente el 65.68% de la variabilidad en la compacidad peor puede ser explicada por los cambios en la dimensión fractal. Este valor sugiere una relación moderada a fuerte entre ambas variables, lo cual es estadísticamente relevante en el contexto del análisis biomédico.
2. Intervalo de confianza
El intervalo de confianza del 95% para el coeficiente de regresión (bi) se encuentra entre 6.6392 y 7.4810, lo que indica que el efecto de la dimensión fractal sobre la compacidad peor es positivo y significativo. Dado que este intervalo no incluye el cero, se confirma la existencia de una relación estadísticamente significativa.
3. Evaluación de supuestos del modelo
Linealidad: La relación entre las variables parece ser lineal, dado el alto valor de R² y la interpretación significativa del coeficiente bi. Sin embargo, no se reportó explícitamente un gráfico de dispersión ni de residuos para confirmarlo visualmente.
4. Normalidad de los residuos: El valor-p de la prueba de Shapiro-Wilk fue 0.0000, lo que indica que los residuos no siguen una distribución normal. Esto viola el supuesto de normalidad, lo cual puede afectar la validez de los intervalos de confianza y pruebas de hipótesis si el tamaño de muestra no es lo suficientemente grande.
5. Homoscedasticidad (igualdad de varianzas): El valor-p de la prueba de Breusch-Pagan también fue 0.0000, indicando que existe heteroscedasticidad, es decir, que los errores no tienen varianza constante. Esto puede llevar a estimaciones ineficientes y errores estándar incorrectos.


Aunque el modelo muestra una relación clara y estadísticamente significativa entre la dimensión fractal y la compacidad peor, no cumple con dos supuestos fundamentales del modelo de regresión lineal clásico: normalidad de los residuos y homoscedasticidad. Esto sugiere que se deben aplicar técnicas de corrección, como transformaciones de variables, regresión robusta o modelos alternativos (e.g., regresión no paramétrica), para mejorar la validez de las inferencias estadístic
